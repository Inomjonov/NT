```python
x = [1, 2, 3, 4, 5]
y = [10, 25, 15, 30, 20]

plt.plot(x, y)
plt.title("Oylik sotuvlar")        # sarlavha
plt.xlabel("Oy")                   # x o'qi nomi
plt.ylabel("Sotuv (ming so'm)")    # y o'qi nomi
plt.grid(True)                     # to'r chiziqlari
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 25, 15, 30, 20]

plt.plot(x, y, color="red", linestyle="--", marker="o", linewidth=2)
plt.title("Bezatilgan grafik")
plt.show()
```

```python
oylar = [1, 2, 3, 4, 5]
matematika = [85, 90, 78, 92, 88]
fizika = [75, 82, 88, 79, 85]

plt.plot(oylar, matematika, color="blue", marker="o", label="Matematika")
plt.plot(oylar, fizika, color="red", marker="s", label="Fizika")

plt.title("Fanlar bo'yicha baholar")
plt.xlabel("Oy")
plt.ylabel("Baho")
plt.legend()          # izohni ko'rsatish
plt.grid(True)
plt.show()
```

```python
fanlar = ["Matem", "Fizika", "Kimyo", "Biologiya"]
baholar = [85, 78, 92, 88]

plt.bar(fanlar, baholar, color="green")
plt.title("Fanlar bo'yicha o'rtacha baho")
plt.xlabel("Fan")
plt.ylabel("Baho")
plt.show()
```

```python
plt.barh(fanlar, baholar, color="lightgreen")
plt.title("Fanlar bo'yicha baho")
plt.show()
```

```python
soat = [1, 2, 3, 4, 5, 6, 7, 8]
baho = [50, 55, 65, 70, 78, 85, 88, 95]

plt.scatter(soat, baho, color="purple", s=100)   # s = nuqta o'lchami
plt.title("O'qish soati va baho bog'liqligi")
plt.xlabel("O'qish soati")
plt.ylabel("Baho")
plt.grid(True)
plt.show()
```

```python
shaharlar = ["Toshkent", "Samarqand", "Buxoro", "Namangan"]
talabalar = [45, 25, 15, 15]

plt.pie(talabalar, labels=shaharlar, autopct="%1.1f%%")
plt.title("Talabalarning shaharlar bo'yicha taqsimoti")
plt.show()
```

```python
import numpy as np

# 200 ta tasodifiy baho
baholar = np.random.normal(75, 10, 200)

plt.hist(baholar, bins=20, color="orange", edgecolor="black")
plt.title("Baholar taqsimoti")
plt.xlabel("Baho")
plt.ylabel("Talabalar soni")
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y1 = [10, 25, 15, 30, 20]
y2 = [5, 15, 25, 10, 30]

plt.figure(figsize=(10, 4))     # oyna o'lchami

plt.subplot(1, 2, 1)            # 1 qator, 2 ustun, 1-o'rin
plt.plot(x, y1, color="blue")
plt.title("Birinchi grafik")

plt.subplot(1, 2, 2)            # 1 qator, 2 ustun, 2-o'rin
plt.bar(x, y2, color="green")
plt.title("Ikkinchi grafik")

plt.tight_layout()              # oraliqni tartibga solish
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 25, 15, 30, 20]

plt.figure(figsize=(10, 8))

plt.subplot(2, 2, 1)
plt.plot(x, y)
plt.title("Chiziqli")

plt.subplot(2, 2, 2)
plt.bar(x, y)
plt.title("Ustunli")

plt.subplot(2, 2, 3)
plt.scatter(x, y)
plt.title("Nuqtali")

plt.subplot(2, 2, 4)
plt.barh(x, y)
plt.title("Gorizontal")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Ma'lumot yaratish
data = {
    "Oy": ["Yanvar", "Fevral", "Mart", "Aprel", "May"],
    "Sotuv": [1200, 1500, 1100, 1800, 2000],
    "Xarajat": [800, 900, 850, 1000, 1100]
}
df = pd.DataFrame(data)

# To'g'ridan-to'g'ri Pandas'dan chizish
df.plot(x="Oy", y="Sotuv", kind="bar", color="teal")
plt.title("Oylik sotuvlar")
plt.ylabel("Summa (ming so'm)")
plt.show()
```

```python
df.plot(x="Oy", y=["Sotuv", "Xarajat"], kind="line", marker="o")
plt.title("Sotuv va xarajat")
plt.ylabel("Summa")
plt.grid(True)
plt.show()
```
