```python
class Talaba:
    def __init__(self, ism, yosh):
        self.ism = ism
        self.yosh = yosh
    
    def salomlash(self):
        print(f"Salom, mening ismim {self.ism}!")

    def malumot(self):
        print(f"Ism: {self.ism}, yosh: {self.yosh}")
```

```python
talaba1 = Talaba('Aziz', 20)
talaba2 = Talaba('Dilnoza', 19)

```

```python
talaba1.salomlash()
talaba2.malumot()
```

```python
class Kalkulyator:
    def __init__(self):
        self.natija = 0
    

    def qoshish(self, son):
        self.natija  += son
        return self.natija

    def ayirish(self, son):
        self.natija -= son
        return self.natija
```

```python
hisobMashina = Kalkulyator()
```

```python
print(hisobMashina.qoshish(10))
```

```python
print(hisobMashina.ayirish(3))
```

```python
class Hayvon:
    def __init__(self, ism):
        self.ism = ism

    def ovqatlanish(self):
        print(f"{self.ism} ovqatlanmoqda!")

```

```python
class Mushuk(Hayvon):
    def miyovlash(self):
        print(f"{self.ism}: miyov")
        
```

```python
mushuk = Mushuk('Tom')
```

```python
mushuk.ovqatlanish()
mushuk.miyovlash()
```

```python
class It:
    def ovoz(self):
        print('Vov-vov')

class Mushuk:
    def ovoz(self):
        print("miyov")

class Sigir:
    def ovoz(self):
        print('Mu-mu')

hayvonlar = [It(), Mushuk(), Sigir()]
for hayvon in hayvonlar:
    hayvon.ovoz()
```

```python
class BankHisob:
    def __init__(self, balans):
        self.__balans = balans #private (yashirin) xususiyat

    def balansni_korsat(self):
        return self.__balans

    def pul_qoshish(self, summa):
        if summa >0:
            self.__balans += summa
```

```python
hisob1  = BankHisob(1000)
hisob1.balansni_korsat()
hisob1.pul_qoshish(2000)
hisob1.balansni_korsat()
```
