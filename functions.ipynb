```python
def salomlash(ism):
    print(f'Salom, {ism}!')
```

```python
salomlash('Messi')
```

```python
def talaba_malumot(ism, yosh, shahar):
    print(f'Ism: {ism}')
    print(f'yosh: {yosh}')
    print(f'shahar: {shahar}')


talaba_malumot('Aziz', 20, 'Namangan')
          
```

```python
def salomlash(ism, salom = "Salom"):
    print(f'{salom}, {ism}')

salomlash('Aziz')
salomlash('Aziz', 'Hayrli kech')
```

```python
# *args
def yigindi(*sonlar):
    natija = 0
    for son in sonlar:
        natija += son
    return natija

print(yigindi(1, 2, 4,5,  12, 34))
```

```python
def eng_katta(*sonlar):
    return max(sonlar)

print(eng_katta(1, 2, 34, 53))
```

```python
# ** kwargs
def malumot(**data):
    print(data)
    for kalit, qiymat in data.items():
        print(f'{kalit}: {qiymat}')

malumot(ism = 'Aziz', yosh = 20, shahar = 'Toshkent')
```
