```python
data = {
    "Ism": ["Aziz", "Dilnoza", "Bobur", "Malika"],
    "Yosh": [20, 19, 21, 20],
    "Baho": [4.5, 4.8, 3.9, 4.2]
}
df = pd.DataFrame(data)
# df

# Bahosi 4.0 dan yuqori talabalar
# past = df[df["Baho"] < 4.0]
# print(past)
# #        Ism  Yosh  Baho
# # 0     Aziz    20   4.5
# # 1  Dilnoza    19   4.8
# # 3   Malika    20   4.2

# Bir nechta shart (& = va, | = yoki)
natija = df[(df["Baho"] > 4.0) & (df["Yosh"] != 20)]
print(natija)
```

```python
import pandas as pd

# 1. Ma'lumot yaratish
data = {
    "Ism": ["Aziz", "Dilnoza", "Bobur", "Malika", "Sardor"],
    "Fan": ["Matem", "Fizika", "Matem", "Fizika", "Matem"],
    "Baho": [85, 92, 78, 88, 95]
}
df = pd.DataFrame(data)

# 2. Excel'ga saqlash
df.to_excel("baholar.xlsx", index=False)

# 3. Qayta o'qish
df = pd.read_excel("baholar.xlsx")

# 4. Filtrlash: 90 dan yuqori
alochi = df[df["Baho"] >= 90]
print("A'lochilar:")
print(alochi)

# 5. O'rtacha baho
print(f"\nO'rtacha baho: {df['Baho'].mean()}")

# 6. CSV'ga saqlash
df.to_csv("baholar.csv", index=False)
print("\nHammasi tayyor!")
```
