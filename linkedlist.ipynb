```python
class Node:
    def __init__(self, qiymat):
        self.qiymat = qiymat
        self.keyingi = None


class BoglanganRoyxat:
    def __init__(self):
        self.bosh = None      # head

    # Boshiga qo'shish — O(1)
    def boshiga_qoshish(self, qiymat):
        yangi = Node(qiymat)
        yangi.keyingi = self.bosh
        self.bosh = yangi

    # Oxiriga qo'shish — O(n)
    def qoshish(self, qiymat):
        yangi = Node(qiymat)
        if self.bosh is None:
            self.bosh = yangi
            return
        joriy = self.bosh
        while joriy.keyingi:
            joriy = joriy.keyingi
        joriy.keyingi = yangi

    # Chiqarish
    def korsatish(self):
        joriy = self.bosh
        natija = []
        while joriy:
            natija.append(str(joriy.qiymat))
            joriy = joriy.keyingi
        print(" -> ".join(natija) + " -> None")

    # Qidirish — O(n)
    def qidirish(self, qiymat):
        joriy = self.bosh
        pozitsiya = 0
        while joriy:
            if joriy.qiymat == qiymat:
                return pozitsiya
            joriy = joriy.keyingi
            pozitsiya += 1
        return -1

    # O'chirish — O(n)
    def ochirish(self, qiymat):
        if self.bosh is None:
            return
        if self.bosh.qiymat == qiymat:
            self.bosh = self.bosh.keyingi
            return
        joriy = self.bosh
        while joriy.keyingi:
            if joriy.keyingi.qiymat == qiymat:
                joriy.keyingi = joriy.keyingi.keyingi
                return
            joriy = joriy.keyingi

    # Uzunlik — O(n)
    def uzunlik(self):
        son = 0
        joriy = self.bosh
        while joriy:
            son += 1
            joriy = joriy.keyingi
        return son

    # Teskari o'girish — O(n)
    def teskari(self):
        oldingi = None
        joriy = self.bosh
        while joriy:
            keyingi = joriy.keyingi
            joriy.keyingi = oldingi
            oldingi = joriy
            joriy = keyingi
        self.bosh = oldingi


# Ishlatish
lst = BoglanganRoyxat()
lst.qoshish(10)
lst.qoshish(20)
lst.qoshish(30)
lst.korsatish()          # 10 -> 20 -> 30 -> None

lst.boshiga_qoshish(5)
lst.korsatish()          # 5 -> 10 -> 20 -> 30 -> None

print(lst.qidirish(20))  # 2
print(lst.uzunlik())     # 4

lst.ochirish(20)
lst.korsatish()          # 5 -> 10 -> 30 -> None

lst.teskari()
lst.korsatish()          # 30 -> 10 -> 5 -> None
```
