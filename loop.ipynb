## while sikli

```python
# shart togri bolgandagina ishlaydi
son = 1
while son <= 5:
    print(son)
    son += 1
    
```

```python
son = 5
for i in range(1, 11):
    print(f"{son} x {i} = {son*i}")
```

```python
for i in range(10):
    if i == 5:
        break
    print(i)
```

```python
for i in range(10):
    if i == 5:
        continue
    print(i)
```
