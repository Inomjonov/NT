## Stack

```python
stek = []

# push — qo'shish
stek.append(1)
stek.append(2)
stek.append(3)
print(stek)          # [1, 2, 3]

# pop — olish (oxirgisi chiqadi)
print(stek.pop())    # 3
print(stek.pop())    # 2
print(stek)          # [1]

# peek — ko'rish (olmasdan)
stek.append(5)
print(stek[-1])      # 5
```

```python
class Stek:
    def __init__(self):
        self.elementlar = []

    def push(self, element):
        self.elementlar.append(element)

    def pop(self):
        if self.bosh_mi():
            return "Stek bo'sh!"
        return self.elementlar.pop()

    def peek(self):
        if self.bosh_mi():
            return "Stek bo'sh!"
        return self.elementlar[-1]

    def bosh_mi(self):
        return len(self.elementlar) == 0

    def hajm(self):
        return len(self.elementlar)


s = Stek()
s.push("A")
s.push("B")
s.push("C")
# print(s.peek())      # C
print(s.pop())       # C
print(s.hajm())      # 2
```

## Queue

```python
from collections import deque

navbat = deque()

# enqueue — qo'shish (oxiriga)
navbat.append("Aziz")
navbat.append("Dilnoza")
navbat.append("Bobur")
# ['Dilnoza', 'Bobur']

# dequeue — olish (boshidan)
print(navbat.popleft())    # Aziz
print(navbat.popleft())    # Dilnoza
print(navbat)              # deque(['Bobur'])
```

```python
import time
from collections import deque

n = 100000

# List bilan — O(n) har safar
royxat = list(range(n))
boshlanish = time.time()
while royxat:
    royxat.pop(0)
print(f"list.pop(0):     {time.time() - boshlanish:.4f} sek")

# Deque bilan — O(1) har safar
dq = deque(range(n))
boshlanish = time.time()
while dq:
    dq.popleft()
print(f"deque.popleft(): {time.time() - boshlanish:.4f} sek")
```
