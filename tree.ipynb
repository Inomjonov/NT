```python
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

```

```python
# Qo'lda shu daraxtni yaratamiz:
#         1
#        / \
#       2   3
#      / \
#     4   5
```

```python
root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)
```

# In-order traversal
4, 2, 5, 1, 3

```python
def inorder(node):
    if node is None:
        return
    inorder(node.left)
    print(node.value, end = ' ')
    inorder(node.right)

inorder(root)
```

# pre-order tugun, chap, o'ng 
# 1, 2, 4, 5, 3

```python
def preorder(node):
    if node is None:
        return
    print(node.value, end = " ")
    preorder(node.left)
    preorder(node.right)

preorder(root)
```

# post-order
# 4, 5, 2, 3, 1

```python
def postorder(node):
    if node is None:
        return
    postorder(node.left)
    postorder(node.right)
    print(node.value, end = ' ')

postorder(root)
```
