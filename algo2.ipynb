```python
import time

n = 100000
royxat = list(range(n))
lugat = {i: i for i in range(n)}
qidirilgan = n - 1

# O(n) — ro'yxatda
boshlanish = time.time()
natija = qidirilgan in royxat
print(f"Ro'yxat (O(n)): {time.time() - boshlanish:.6f} sek")

# O(1) — lug'atda
boshlanish = time.time()
natija = qidirilgan in lugat
print(f"Lug'at (O(1)):  {time.time() - boshlanish:.6f} sek")

# Lug'at minglab marta tezroq!
```

```python
# Ro'yxatdan yig'indisi berilgan songa teng bo'lgan juftlikni toping
def ikki_son(sonlar, maqsad):
    korilgan = {}
    for i, son in enumerate(sonlar):
        kerakli = maqsad - son
        if kerakli in korilgan:          # O(1) tekshirish!
            return [korilgan[kerakli], i]
        korilgan[son] = i
    return None

print(ikki_son([2, 7, 11, 15], 9))    # [0, 1]  -> 2 + 7 = 9
```

```python
# # Set yaratish
sonlar = {1, 2, 3, 4, 5}
# bosh = set()

# # Takrorlarni olib tashlash
# royxat = [1, 2, 2, 3, 3, 3, 4]
# noyob = set(royxat)
# print(noyob)              # {1, 2, 3, 4}
# print(list(noyob))        # [1, 2, 3, 4]

# Qo'shish va o'chirish
sonlar.add(6)
sonlar.discard(1)
print(sonlar)             # {2, 3, 4, 5, 6}

# Qidirish — O(1)!
print(3 in sonlar)        # True
```

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

# print(a | b)     # {1, 2, 3, 4, 5, 6}  -> birlashma
# print(a & b)     # {3, 4}              -> kesishma
# print(a - b)     # {1, 2}              -> ayirma
print(a ^ b)     # {1, 2, 5, 6}        -> simmetrik ayirma
```

```python
# Ikki guruhda ham qatnashgan talabalar
matematika = {"Aziz", "Dilnoza", "Bobur"}
fizika = {"Dilnoza", "Bobur", "Malika"}

print(matematika & fizika)    # {'Dilnoza', 'Bobur'}  -> ikkalasida
print(matematika - fizika)    # {'Aziz'}              -> faqat matematikada
print(matematika | fizika)    # hammasi
```
